Lakukan proses standardisasi pada semua kolom yang memiliki nilai numerik.

```python
from sklearn.preprocessing import StandardScaler


# X berisi 30 kolom numerik murni
X = df.drop(columns=["id", "diagnosis", "Unnamed: 32"], errors="ignore")

# 2. Ambil daftar kolom numerik dari X saja
kolom_numerik = X.select_dtypes(include=["int64", "float64"]).columns

# 3. Lakukan standardisasi pada X
scaler = StandardScaler()
X_scaled = X.copy()
X_scaled[kolom_numerik] = scaler.fit_transform(X[kolom_numerik])

# Tampilkan hasil
print("Kolom yang distandardisasi (Total 30):")
print(kolom_numerik.tolist())

print("\nHasil Standardisasi (X_scaled):")
print(X_scaled.head())
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
path = "/content/drive/MyDrive/Pembelajaran mesin/voice.csv"

df = pd.read_csv(path)

print("Ukuran dataset:", df.shape)
df.head()
```

2. Pisahkan fitur dan label

```python
X = df.iloc[:, :-1]
y = df.iloc[:, -1]

print("Jumlah fitur:", X.shape[1])
print("Jumlah data:", X.shape[0])
print("Kelas:", y.unique())
```

3. Standardisasi data

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

split svm 70:30

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

svm_linear_70 = SVC(kernel='linear')
svm_linear_70.fit(X_train, y_train)

y_pred = svm_linear_70.predict(X_test)

acc_linear_70 = accuracy_score(y_test, y_pred)

print("Akurasi SVM Linear 70:30:", acc_linear_70)
```

SVM kernel Polynomial

```python
svm_poly_70 = SVC(kernel='poly')

svm_poly_70.fit(X_train, y_train)

y_pred = svm_poly_70.predict(X_test)

acc_poly_70 = accuracy_score(y_test, y_pred)

print("Akurasi SVM Polynomial 70:30:", acc_poly_70)
```

SVM kernel RBF

```python
svm_rbf_70 = SVC(kernel='rbf')

svm_rbf_70.fit(X_train, y_train)

y_pred = svm_rbf_70.predict(X_test)

acc_rbf_70 = accuracy_score(y_test, y_pred)

print("Akurasi SVM RBF 70:30:", acc_rbf_70)
```

split 80:20

```python
def evaluasi_svm(X, y, test_size, kernel):
    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=test_size,
        random_state=42,
        stratify=y
    )

    model = SVC(kernel=kernel)
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    return accuracy
```

jalankan semua kombinasi:

```python
hasil = []

for split in [0.30, 0.20]:
    for kernel in ['linear', 'poly', 'rbf']:
        acc = evaluasi_svm(
            X_scaled,
            y,
            test_size=split,
            kernel=kernel
        )

        rasio = f"{int((1-split)*100)}:{int(split*100)}"

        hasil.append({
            "Split": rasio,
            "Kernel": kernel,
            "Accuracy": acc
        })

hasil_df = pd.DataFrame(hasil)

hasil_df
```

```python
hasil_df["Accuracy (%)"] = hasil_df["Accuracy"] * 100

hasil_df
```

Mencari model terbaik

```python
terbaik = hasil_df.loc[hasil_df["Accuracy"].idxmax()]

print("Model terbaik:")
print(terbaik)
```

2. Membuat model klasifikasi male dan female menggunakan SVM dengan kernel RBF menggunakan data voice.csv, rasio 80:20, kemudian melakukan tuning hyperparameter dan mencatat akurasinya.

```python
path = "/content/drive/MyDrive/Pembelajaran mesin/voice.csv"

df = pd.read_csv(path)

X = df.iloc[:, :-1]
y = df.iloc[:, -1]

print("Ukuran data:", df.shape)
print("Kelas:", y.unique())
```

Split 80:20

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

Standardisasi

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

SVM RBF

```python
svm_rbf = SVC(kernel='rbf')

svm_rbf.fit(X_train, y_train)

y_pred = svm_rbf.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Akurasi SVM RBF:", accuracy)
print("Akurasi (%):", accuracy * 100)
```

Tuning C dan gamma

```python
hasil_tuning = []

for C in [0.1, 1, 10, 100]:
    for gamma in ['scale', 0.001, 0.01, 0.1, 1]:

        model = SVC(
            kernel='rbf',
            C=C,
            gamma=gamma
        )

        model.fit(X_train, y_train)

        y_pred = model.predict(X_test)

        acc = accuracy_score(y_test, y_pred)

        hasil_tuning.append({
            'C': C,
            'Gamma': gamma,
            'Accuracy': acc
        })

tuning_df = pd.DataFrame(hasil_tuning)

tuning_df['Accuracy (%)'] = (
    tuning_df['Accuracy'] * 100
).round(2)

tuning_df
```

Cari yang terbaik:

```python
best = tuning_df.loc[tuning_df['Accuracy'].idxmax()]

print("Hasil terbaik:")
print(best)
```
